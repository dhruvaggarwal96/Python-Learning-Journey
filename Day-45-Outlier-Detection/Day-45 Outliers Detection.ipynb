```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns


df = pd.read_csv("employee_data.csv")
df.head()
```

## Boxplot

```python
sns.boxplot(x = df["Salary"])
plt.show()
```

### IQR 

```python
Q1 = df["Salary"].quantile(0.25)
Q3 = df["Salary"].quantile(0.75)

IQR = Q3 - Q1 
lower = Q1 - (1.5 * IQR)
upper = Q3 + (1.5 * IQR)
print(lower , upper ) 
```

```python
df[(df["Salary"] > lower) | (df["Salary"] < upper ) ] 
df[(df["Salary"] >= lower) & (df["Salary"] <= upper ) ] 
```
