## Day 26 — Pie Charts 🥧

A Pie Chart is used to show proportions (parts of a whole).

Each slice represents a percentage of the total.

```python
import matplotlib.pyplot as plt

labels = ["Food" , "Rent" , "Travel" , "Shopping"]
expenses = [40 , 30 , 15, 15]

plt.pie(expenses , labels = labels)
plt.show()
```

### Exercise 1 : Monthly Expenses 

```python
import matplotlib.pyplot as plt 

labels = ["Food", "Rent", "Travel", "Shopping"]
expenses = [40, 30, 15, 15]

plt.pie(expenses , labels = labels , autopct = "%1.1f%%")
plt.title("Montly Expenses")
plt.show()
```

Food

Travel and Shopping

### E-commerce Sales Distribution



```python
categories = [
    "Electronics",
    "Clothing",
    "Furniture",
    "Beauty",
    "Books"
]

sales = [450000, 250000, 150000, 100000, 50000]
```

#### Tasks:

- Create a pie chart.
- Calculate the percentage contribution of each category.
- Which categories contribute to more than 20%?
- Which category contributes less than 10%?

```python
plt.pie(sales , labels = categories , autopct = "%1.1f%%")
plt.title("E-commerce Sales Distribution")
plt.show()
```

```python
total = sum(sales)
sales_percentage = []
for i in sales:
    sales_percentage.append((i/total)*100)
# print(sales_percentage)
for i in sales_percentage:
    if i > 20:
        index = sales_percentage.index(i)
        print("categories contribute to more than 20% is :", categories[index])
for i in sales_percentage:
    if i < 10:
        index = sales_percentage.index(i)
        print("category contributes less than 10% is : " , categories[index])
```
