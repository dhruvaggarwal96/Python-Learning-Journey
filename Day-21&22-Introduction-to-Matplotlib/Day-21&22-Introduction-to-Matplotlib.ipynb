```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

plt.plot(x,y)

plt.title("Sample Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")

plt.show()
```

```python
import matplotlib.pyplot as plt

students = ["A" , "B" , "C" , "D" , "E"]
marks = [89 , 90 , 97 , 76 , 56 ]

plt.plot(students , marks)

plt.title("Student Marks")
plt.xlabel("Students")
plt.ylabel("Marks")

plt.show()
```

```python
# Monthly Expenses
import matplotlib.pyplot as plt

months = ["Jan" , "Feb" , "Mar" , "Apr" , "May" , "Jun"]
expenses = [5000 , 7000 , 6000 , 8000 , 9000 , 10000]

plt.plot(months , expenses)

plt.totle("Monthly Expenses")
plt.xlabel("Months")
plt.ylabel("Expenses")

plt.show()
```

```python
# Daily Study Hours

days = ["Mon" , "Tue" , "Wed" , "Thu" , "Fri"]
study_hours = [8 , 6 , 7 , 5 , 9]

plt.plot(days , study_hours)

plt.title("Study Hours")
plt.xlabel("Days")
plt.ylabel("Study Hours")

plt.show()
```

```python
# Temperature Graph

temperatures = [30,32,35,33,31,29,28]
days = ["Mon","Tue","Wed","Thu","Fri","Sat","Sun"]

plt.plot(days , temperatures)

plt.title("Temperature Graph")
plt.xlabel("Days")
plt.ylabel("Temperatures")

plt.show()
```

```python
sales = [1200,1500,1800,1700,2200]
products = ["A","B","C","D","E"]

plt.plot(products , sales)

plt.title("Sales Graph")
plt.xlabel("Products")
plt.ylabel("Sales")

plt.show()
```

```python
# Changing the Line Color
plt.plot(products , sales , color = "red")

plt.title("Sales Graph")
plt.xlabel("Products")
plt.ylabel("Sales")

plt.show()
```

```python
# changing the line style
plt.plot(products , sales , linestyle = "dashed")

plt.title("Sales Graph")
plt.xlabel("Products")
plt.ylabel("Sales")

plt.show()
```

```python
# Adding Markers
plt.plot(products , sales , linestyle = "dashed" , marker = "o")

plt.title("Sales Graph")
plt.xlabel("Products")
plt.ylabel("Sales")

plt.show()
```

```python
# Question 1

# Create a chart:

# days = [1,2,3,4,5]
# water = [2,3,4,3,5]

# Requirements:

# Blue line
# Circle marker
# Title
# Labels

days = [1,2,3,4,5]
water = [2,3,4,3,5]

plt.plot(days , water , color = "blue" , marker = "o" )

plt.title("Water Consumption")
plt.xlabel("Days")
plt.ylabel("Water Consumption")

plt.show()
```

```python
# Create a chart:

# hours = [2,4,3,5,6]
# days = ["Mon","Tue","Wed","Thu","Fri"]

# Requirements:

# Green line
# Star marker
# Grid

hours = [2,4,3,5,6]
days = ["Mon","Tue","Wed","Thu","Fri"]
plt.plot(days , hours , color = "green" , marker = "*")

plt.title("Study Hours")
plt.xlabel("Days")
plt.ylabel("Study Hours")

plt.grid()

plt.show()
```

```python
# Create a chart:

months = ["Jan","Feb","Mar","Apr","May"]
profit = [1000,1200,1500,1300,1800]

# Requirements:

# Red line
# Dashed style
# Grid
# Title

plt.plot(months , profit , color = "red" , linestyle = "dashed")

plt.title("Profit Graph")
plt.xlabel("Months")
plt.ylabel("Profit")

plt.grid()

plt.show()
```

```python
# Create a chart showing:

temperature = [30,31,33,35,34,32,29]
days = ["Mon","Tue","Wed","Thu","Fri","Sat","Sun"]

# Requirements:

# Marker
# Grid
# Labels
# Title

plt.plot(days , temperature , marker = "o")

plt.title("temperature Graph")

plt.xlabel("Days")
plt.ylabel("Temperature")
plt.grid()
plt.show()
```

```python
# multiple lines in one graph
import matplotlib.pyplot as plt

days = [1,2,3,4,5]

sales = [100,120,150,170,200]
profit = [20,30,40,45,60]

plt.plot(days, sales , label = "Sales" )
plt.plot(days, profit , label = "Profit" )

plt.title("Sales and Profit")
plt.xlabel("Days")
plt.ylabel("Values")

plt.legend()

plt.show()
```

```python
# Sales vs Profit

import matplotlib.pyplot as plt

days = [1,2,3,4,5]

sales = [100,120,150,170,200]
profit = [20,30,40,45,60]

plt.plot(days , sales ,marker = "o" ,label = "Sales" )

plt.plot(days , profit ,
         label = "Profit" ,
         marker = "o" ,
         )

plt.title("Sales and Profit")
plt.xlabel("Days")
plt.ylabel("Values")

plt.legend()
plt.grid()
plt.show()
```

```python
views = [1000,1200,1500,1300,2000,2500,3000]

days = ["Mon","Tue","Wed","Thu","Fri","Sat","Sun"]
# Task

# Create a chart and answer:

# Which day had maximum views?
# Which day had minimum views?
# Average views?

plt.plot(days , views , marker ="o")

plt.title("Views Graph")
plt.xlabel("Days")
plt.ylabel("Views")
plt.grid()
plt.show()
# Which day had maximum views?
print(max(views))

# Which day had minimum views?
print(min(views))
# Average views?
avg = sum(views) / len(views)
print(avg)
```
