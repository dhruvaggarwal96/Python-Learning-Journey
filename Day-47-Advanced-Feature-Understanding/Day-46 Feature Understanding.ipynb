```python
# columns that have at least one missing value.
missing = df.isnull().sum() > 0
missing[missing].index
```

```python
#Column name
# Missing value count
# Missing value percentage
# for only those columns that have missing values, sorted from highest percentage to lowest.

missing_count = df.isnull().sum() 
columns = missing_count.index
missing_per = (missing_count / len(df) )*100
df_missing = pd.DataFrame({"Columns_Name" : columns , "Missing_count" : missing_count ,"Missing_percentage" :  missing_per})
df_missing[df_missing["Missing_count"] > 0].sort_values(ascending = False , by = "Missing_percentage")
```

```python
# top 10 columns with the highest number of unique values.
df.nunique().sort_values(ascending = False).head(10)
```

```python
# all columns that have only one unique value.
unique = df.nunique() == 1
unique[unique].index
```

```python
# all categorical (object) columns and display how many unique values each categorical column has.
columns = df.select_dtypes(include = ["object"]).columns
unique_count = df.select_dtypes(include = ["object"]).nunique()
df_unique = pd.DataFrame({"Column Name" : columns , "Unique Count" : unique_count})
df_unique.sort_values(ascending = False , by = "Unique Count")
```

```python
# all categorical (object) columns that have more than 10 unique values.
object_df = df.select_dtypes(include = ["object"])
greater_10 = object_df.nunique() > 10
greater_10[greater_10].index
```

```python
# all numerical columns that have more than 100 unique values.
df_number = df.select_dtypes(include = ["number"])
num_gt_100 = df_number.nunique() > 100
num_gt_100[num_gt_100].index
```

```python
# summary table for all numerical columns containing:
# Column Name
# Data Type
# Unique Values
# Missing Count
df_number = df.select_dtypes(include = ["number"])
Data_Types = df_number.dtypes
unique = df_number.nunique()
missing = df_number.isnull().sum()
df_numbers = pd.DataFrame({"Data Type" : Data_Types , "Unique Values" : unique , "Missing Count" : missing})
print(df_numbers)
```

```python
# summary table for all categorical columns containing:

# Column Name
# Data Type
# Unique Values
# Missing Count
# Missing Percentage
df_object = df.select_dtypes(include = ["object"])
columns = df_object.columns
d_type = df_object.dtypes
unique = df_object.nunique()
missing = df_object.isnull().sum()
missing_percentage = (( missing / len(df_object) ) * 100 ).round(2)
df_objects = pd.DataFrame({"Column Name" : columns , "Data Type" : d_type , "Unique Values" : unique , "Missing Count" : missing , "Missing Percentage" : missing_percentage})
df_objects.sort_values(ascending = False , by = "Missing Percentage")
```

```python
# summary table for all columns (numerical + categorical) containing:

# Column Name
# Data Type
# Unique Values
# Missing Count
# Missing Percentage
# Cardinality (High / Low)
df_num_obj = df.select_dtypes(include = ["object" , "number"])
columns = df_num_obj.columns
d_type = df_num_obj.dtypes
unique = df_num_obj.nunique()
missing = df_num_obj.isnull().sum()
missing_percentage = ((missing / len(df_num_obj))*100).round(2) 
df_num_objs = pd.DataFrame({"Column Name" : columns , "Data Type" : d_type ,
                            "Unique Values" : unique , "Missing Count" : missing , 
                            "Missing Percentage" : missing_percentage})
df_num_objs["Cardinality"] = np.where(df_num_objs["Unique Values"] > 10 , "High" , "Low")
df_num_objs
```
