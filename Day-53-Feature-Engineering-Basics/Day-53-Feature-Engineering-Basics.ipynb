```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('hotel_bookings.csv')
df.head()
```

```python
df['total_guests'] = df['adults'] + df['children'] + df['babies']
df
```

```python
df['stay_duration'] = df['stays_in_weekend_nights'] + df['stays_in_week_nights']
df
```

```python
df['total_revenue'] = df['adr'] * df["stay_duration"]
df
```

```python
df['is_family'] = (df['children'] + df['babies']) > 0
df
```

```python
df['revenue_per_guest'] = np.where(df['total_guests'] == 0 ,np.nan ,df['total_revenue'] / df['total_guests']) 
df
```

```python
df['is_long_stay'] = np.where(df['stay_duration'] >= 7 , True , False)
df
```

```python
df['lead_time_group'] = pd.cut(df['lead_time'] , labels = ['Short' , 'Medium' , 'Long' , 'Very Long'] ,bins =  [-1,30,90,180,df['lead_time'].max()])
df
```

```python
df['weekend_ratio'] = np.where(df['stay_duration'] == 0 ,np.nan,df['stays_in_weekend_nights'] / df['stay_duration'])
```

```python
df['has_special_request'] = np.where(df['total_of_special_requests'] > 0 , True , False)
```

```python
df['guest_type'] = pd.cut(df['total_guests'] , bins = [-1,1,2,4,df['total_guests'].max()] , labels = ["Solo" , "Couple" , 'Family' , 'Group'])

# OR
conditions = [df['total_guests'] == 1 , 
              df['total_guests'] == 2 , 
             df['total_guests'].between(3,4),
             df['total_guests'] >= 5]

choices = ["Solo" , "Couple" , "Family" , "Group"]
df['guest_type'] = np.select(conditions , choices , default = "Unknown")
```

```python
df['high_revenue_booking'] = np.where(df['total_revenue'] > df['total_revenue'].mean() , True , False)
```

```python
conditions = [df['arrival_date_month'].isin(["December" , "January" , "February"]),
             df['arrival_date_month'].isin(["March" , "April" , "May"]),
             df['arrival_date_month'].isin(["June" , "July" , "August"]),
             df['arrival_date_month'].isin(["September" , "October" , "November"])]
choices = ["Winter" , "Spring" , "Summer" , "Autumn"]
df["season"] = np.select(conditions , choices , default = "Unknown")
```

```python
conditions = [(df["lead_time"] >= 180) & (df["is_repeated_guest"] == 0) , 
              (df["lead_time"] >= 180) & (df["is_repeated_guest"] == 1) ,]
choices = ['High Risk' , 'Loyal Planner']
df["booking_priority"] = np.select(conditions , choices , default = "Normal")
              
```

```python
conditions = [(df["total_revenue"] >= 1000) & (df['total_guests'] >= 4) , 
              (df['total_revenue'] >= 500 )]
choices = ['Premium' , 'Standard']
df['booking_value'] = np.select(conditions , choices , default = 'Budget')
```

```python
conditions = [df["is_canceled"] == 1,
             df['booking_changes'] > 2 , 
             df['is_repeated_guest'] == 1]
choices = ['Cancelled' , 'Modified' , 'Repeat']
df['booking_status'] = np.select(conditions , choices , default = "Confirmed")
```
