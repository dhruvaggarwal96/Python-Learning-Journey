```python
import pandas as pd
import numpy as np 
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('hotel_bookings.csv')
df.head()
```

```python
# Hypothesis:
# Customers with longer lead times are more likely to cancel.

# Code
df.groupby("is_canceled")["lead_time"].mean()
```

```python
# Observation:
# Canceled bookings have an average lead time of 144.85 days,
# while non-canceled bookings have an average lead time of 79.98 days.

# Conclusion:
# Hypothesis Accepted.
```

```python
# Hypothesis:
# Resort Hotel guests spend more per booking than City Hotel guests.

# Code
df.groupby("hotel")["adr"].mean()
```

```python
# Observation:
# The average ADR for City Hotel is 105.30,
# whereas the average ADR for Resort Hotel is 94.95.

# Conclusion:
# Hypothesis Rejected.

# Business Insight:
# City Hotels generate higher average room revenue per booking than Resort Hotels.
# This suggests that City Hotels may attract higher-paying guests or have stronger pricing
# due to business travel, urban demand, or premium city locations.
```

```python
# Hypothesis:
# Repeated guests are less likely to cancel their bookings.

# Code
df.groupby('is_repeated_guest' , as_index = True)['is_canceled'].mean()
```

```python
# Observation:
# New guests have a cancellation rate of 37.79%,
# while repeated guests have a much lower cancellation rate of 14.49%.

# Conclusion:
# Hypothesis Accepted.

# Business Insight:
# Repeated guests are more loyal and significantly less likely to cancel their bookings.
# Hotels can improve retention by encouraging first-time guests to become repeat customers
# through loyalty programs, personalized offers, and membership benefits.
```

```python
# Hypothesis
# Customers with children spend more on hotel bookings than customers without children.

# Code
df['have_children'] = df['children'] > 0
df.groupby('have_children')['adr'].mean()
```

```python
# Observation:
# Customers without children have an average ADR of 97.47,
# while customers with children have a significantly higher average ADR of 158.08.

# Conclusion:
# Hypothesis Accepted.

# Business Insight:
# Families traveling with children spend considerably more per booking than guests without children.
# Hotels can maximize revenue by offering family packages, kid-friendly amenities,
# and targeted promotions to attract more family travelers.
```

```python
# Hypothesis
# Customers who requested parking stay longer than those who did not request parking.

# Code
df['total_stay'] = df['stays_in_week_nights'] + df['stays_in_weekend_nights']
df['req_parking_Truly'] = df['required_car_parking_spaces'] > 0
df.groupby('req_parking_Truly')['total_stay'].mean()
```

```python
# Observation:
# Guests who did not request parking stayed for an average of 3.45 nights,
# whereas guests who requested parking stayed for an average of 3.16 nights.

# Conclusion:
# Hypothesis Rejected.

# Business Insight:
# Requesting a parking space does not indicate a longer hotel stay.
# Parking demand may be influenced by guests arriving in personal vehicles
# rather than the duration of their stay.
```

```python
# Hypothesis
# Customers with previous cancellations are more likely to cancel their current booking.

# Code
df['has_previous_cancellation'] = df['previous_cancellations'] > 0
df.groupby('has_previous_cancellation')['is_canceled'].mean() * 100
```

```python
# Observation:
# Customers with no previous cancellations have a current cancellation rate of 33.91%,
# whereas customers with previous cancellations have a much higher cancellation rate
# of 91.64%.

# Conclusion:
# Hypothesis Accepted.

# Business Insight:
# Previous cancellation history is a strong indicator of future cancellations.
# Hotels should monitor customers with prior cancellations and consider
# stricter booking policies, advance payments, or personalized confirmation
# reminders to reduce cancellation risk.
```

```python
# Hypothesis
# Bookings with special requests have a lower cancellation rate than bookings without special requests.

# Code
df['has_special_request'] = df['total_of_special_requests'] > 0
df.groupby('has_special_request')['is_canceled'].mean()
```

```python
# Observation:
# Bookings without special requests have a cancellation rate of 47.72%,
# while bookings with at least one special request have a much lower
# cancellation rate of 21.74%.

# Conclusion:
# Hypothesis Accepted.

# Business Insight:
# Guests who make special requests are more committed to their trips and
# are less likely to cancel. Hotels can encourage guests to personalize
# their bookings, as this may improve booking commitment and reduce
# cancellation rates.
```

```python
# Hypothesis
# Bookings with deposit payments are less likely to be canceled than bookings with no deposit.

# Code
df.groupby('deposit_type')['is_canceled'].mean() * 100
```

```python
# Observation:
# Bookings with 'No Deposit' have a cancellation rate of 28.38%.
# 'Refundable' bookings have a cancellation rate of 22.22%.
# 'Non Refund' bookings have an extremely high cancellation rate of 99.36%.

# Conclusion:
# Hypothesis Rejected.

# Business Insight:
# Although refundable deposits have a lower cancellation rate than no-deposit bookings,
# non-refundable bookings have an exceptionally high cancellation rate in this dataset.
# This suggests the relationship between deposit type and cancellations is more complex
# and should be investigated further before making business decisions.
```

```python
# Hypothesis
# Guests staying on weekends have a higher average ADR than guests staying only on weekdays.

# Code
df['weekend_nights_gt_1'] = df['stays_in_weekend_nights'] > 0
df.groupby('weekend_nights_gt_1')['adr'].mean()
```

```python
# Observation:
# Guests staying only on weekdays have an average ADR of 98.19,
# while guests staying on at least one weekend night have an average ADR of 104.64.

# Conclusion:
# Hypothesis Accepted.

# Business Insight:
# Weekend stays generate higher average room revenue than weekday-only stays.
# Hotels can maximize revenue by promoting weekend packages,
# premium weekend experiences, and dynamic pricing during weekends.
```

```python
# Hypothesis
# Bookings made through Online Travel Agents (OTA) have a higher cancellation rate than bookings made through Direct bookings.

# Code
df.groupby('distribution_channel')['is_canceled'].mean() 
```

```python
# Observation:
# Direct bookings have a cancellation rate of 17.46%.
# TA/TO (Travel Agents / Tour Operators) bookings have a much higher
# cancellation rate of 41.03%.
# Corporate and GDS channels also have higher cancellation rates than Direct.
# The Undefined category has the highest cancellation rate (80%), but it should
# be interpreted carefully because it may contain very few records.

# Conclusion:
# Hypothesis Accepted.

# Business Insight:
# Bookings made through travel agents have a substantially higher cancellation
# rate than direct bookings. Hotels can reduce cancellations by encouraging
# customers to book directly through loyalty programs, exclusive discounts,
# or additional booking benefits.
```
