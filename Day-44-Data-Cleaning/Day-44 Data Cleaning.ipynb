```python
import pandas as pd 

df = pd.read_csv("employee_data.csv") 
df.head()
```

```python
print("Number of Rows : " , df.shape[0] ) 
print("Number of Columns : " , df.shape[1] ) 
```

```python
df.info()
```

```python
df.isnull().sum()
```

#### Find the percentage of missing values in each column.

```python
( df.isnull().sum() / len(df) ) * 100
```

#### Find the total number of duplicate rows in the dataset.

```python
df.duplicated().sum()
```

#### Remove all duplicate rows from the original DataFrame.

```python
df.drop_duplicates(inplace = True)
```

#### Display the names of all columns in the DataFrame.

```python
df.columns
```

#### Rename the column Date_Joined to Joining_Date.

```python
df.rename( columns = { "Date_Joined" : "Joining_Date"} , inplace = True) 
```

#### Convert the Joining_Date column to the datetime data type.

```python
df["Joining_Date"] = pd.to_datetime(df["Joining_Date"] , errors = "coerce")
```

#### Find all rows where Age is missing.

```python
df[df["Age"].isnull()]
```

#### Find all rows where the Salary column has missing values.

```python
df[df["Salary"].isnull()]
```

#### Find all rows where Age is not missing.

```python
df[df["Age"].notnull()]
```

#### Find all rows where Age is greater than 30.

```python
df[df["Age"] > 30]
```

#### Find all rows where Department is "IT".

```python
df[df["Department"] == "IT"]
```

#### Find all rows where Salary is greater than 60000 AND Department is "IT".

```python
df[(df["Salary"] > 60000) & (df["Department"] == "IT")]
```

#### Find all rows where Department is either "IT" or "HR".

```python
df[(df["Department"] == "IT") | (df["Department"] == "HR")]
```

#### Find all rows where Salary is missing OR Age is missing.

```python
df[(df["Salary"].isnull()) | (df["Age"].isnull())]
```

#### Find all rows where Age is not numeric.

```python
df[pd.to_numeric(df["Age"] , errors = "coerce").isna()]
```

#### Find all rows where Joining_Date cannot be converted into a valid datetime.

```python
df[pd.to_datetime(df["Joining_Date"] , errors = "coerce").isna()]
```
