Day 17 – Selecting & Filtering Data (Pandas)
🎯 Today's Goal

By the end of today, you should be able to:

✅ Select single columns

✅ Select multiple columns

✅ Filter rows based on conditions

✅ Apply multiple conditions

✅ Find specific data from a dataset

📚 Study Notes
1️⃣ Selecting a Single Column
df["Sales"]

Example:

import pandas as pd

df = pd.read_csv("superstore.csv")

print(df["Sales"])
2️⃣ Selecting Multiple Columns
df[["Sales", "Profit"]]

Example:

print(df[["Sales", "Profit"]])
3️⃣ Filtering Data
Sales greater than 1000
df[df["Sales"] > 1000]

Example:

high_sales = df[df["Sales"] > 1000]

print(high_sales)
4️⃣ Multiple Conditions
Sales > 1000 AND Profit > 100
df[(df["Sales"] > 1000) & (df["Profit"] > 100)]

Example:

result = df[(df["Sales"] > 1000) & (df["Profit"] > 100)]

print(result)

```python
import pandas as pd

df = pd.read_csv("customer_dataset.csv")
df.head()
```

```python
# Display only the Name column.
df["Name"]
```

```python
# Display only the City column.
df["City"]
```

```python
# Display both Name and PurchaseAmount.
df[["Name" , "PurchaseAmount"]]
```

```python
# Find customers whose PurchaseAmount > 2000.
df[df["PurchaseAmount"] > 2000]
```

```python
# Find customers from Delhi.
df[df["City"] == "Delhi"]
```

```python
# Find customers whose Age > 25.
df[df["Age"] > 20]
```

```python
# Find customers from Delhi whose purchase amount is greater than 1000.
df[(df["City"] == "Delhi") & (df["PurchaseAmount"] > 1000)]
```

```python
# Find customers whose age is less than 30.
df[df["Age"] < 30]
```

```python
# Find customers whose purchase amount is between 1500 and 3000.
df[(df["PurchaseAmount"]<3000) & (df["PurchaseAmount"]>1500)]
```

```python
# Find customers whose city is either Delhi or Mumbai.
df[(df["City"] == "Delhi") & (df["City"] == "Mumbai")]
```

```python
# Find customers with missing values.
customer_with_missing = df[df.isnull().any(axis=1)]
customer_with_missing
```

```python
# Find customers whose age is missing.
df[df["Age"].isnull()]
```

```python
# Find the top 3 customers with the highest purchase amount.
top_3_customer = df.sort_values(by = "PurchaseAmount" , ascending = False).head(3)
top_3_customer
```

```python
avg_purchase_amt = df["PurchaseAmount"].mean()
df[df["PurchaseAmount"] > avg_purchase_amt]
```

```python
# Find the oldest customer.
k = df["Age"].max()
k
df[df["Age"] == k]
```

```python
# Find customers whose purchase amount is below 2000 and age is above 25.
df[(df["PurchaseAmount"] < 2000) & (df["Age"] > 25)]
```

```python
import pandas as pd

df = pd.read_csv("sales_data.csv")
```

```python
# Display only Sales_Amount.
df["Sales_Amount"]
```

```python
# Display Product_Category and Sales_Amount.
df[["Product_Category" , "Sales_Amount"]]
```

```python
# Find sales where Sales_Amount > 5000.
df[df["Sales_Amount"] > 5000]
```

```python
# Find sales made in the North region.
df[df["Region"] == "North"]
```

```python
# Find all online sales
df[df["Sales_Channel"] == "Online"]
```

```python
# Find all retail sales.
df[df["Sales_Channel"] == "Retail"]
```

```python
# Find customers with discounts greater than 10.

df[df["Discount"] > 10]
```

```python
# Find sales where:
# Sales_Amount > 5000
# AND
# Quantity_Sold > 5

df[(df["Sales_Amount"] > 5000) & (df["Quantity_Sold"] > 5)]
```

```python
# Find sales made by a specific sales rep.
df[df["Sales_Rep"] == "Bob"]
```

```python
# Find sales from the South region.
df[df["Region"] == "South"]
```

```python
# Find sales with discounts greater than 20%.
df[df["Discount"]  > 20 ]
```

```python
# Find sales paid using a specific payment method.
df[df["Payment_Method"] == "Bank Transfer"]
```

```python
# Find products whose unit price is greater than 500.
df[df["Unit_Price"] > 500]
```

```python
# Find B2B customers.
df[df["Customer_Type"] == "B2B"]
```

```python
# Find B2C customers.
df[df["Customer_Type"] == "B2C"]
```

```python
# Find the top 10 highest sales amounts.
df.sort_values(by = "Sales_Amount" , ascending = False).head(10)
```

```python
# Find the top 5 most expensive products (Unit_Price)
df.sort_values(by = "Unit_Price" , ascending = False).head(5)
```

```python
# Find products with the highest quantity sold.
df.sort_values(by = "Quantity_Sold" , ascending = False).head(5)
```

```python
# Find online sales with discounts greater than 15%.
df[(df['Sales_Channel'] == "Online") & (df["Discount"] > 15)]
```

```python
# Find sales from North region AND Online channel.
df[(df['Sales_Channel'] == "Online") & (df["Region"] == "North")]
```

```python
# Find products whose sales amount is above the average sales amount.
df[df["Sales_Amount"] > df["Sales_Amount"].mean()]
```
