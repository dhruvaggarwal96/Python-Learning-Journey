## Exploratory Data Analysis

### 10.1 Hotel Type Distribution

Understanding the distribution of bookings across different hotel types helps identify customer preferences and the relative demand for each hotel category. This information can support business planning, pricing strategies, and resource allocation.

**Business Question**

Which hotel type receives more bookings?

```python
# Number of bookings by hotel type
hotel_distribution = df["hotel"].value_counts()

hotel_distribution
```

```python
plt.figure(figsize=(6,4))

sns.countplot(data=df, x="hotel")

plt.title("Distribution of Hotel Bookings")
plt.xlabel("Hotel Type")
plt.ylabel("Number of Bookings")

plt.show()
```

### Observation

The distribution shows how bookings are divided between City Hotels and Resort Hotels. Comparing the booking counts helps identify which hotel type attracts more customers.

### Business Insight

If one hotel type consistently receives a higher number of bookings, management can investigate the factors driving its demand and use these insights to optimize pricing, marketing strategies, and operational planning for both hotel types.

### 10.2 Country-wise Booking Analysis

Understanding the geographic distribution of hotel bookings helps identify the hotel's most valuable customer markets. This analysis can support marketing strategies, international promotions, and customer acquisition efforts.

**Business Question**

Which countries contribute the highest number of hotel bookings?

```python
# Top 10 countries by number of bookings
top_countries = (
    df["country"]
    .value_counts()
    .head(10)
)

top_countries
```

```python
plt.figure(figsize=(10,5))

ax = sns.barplot(
    x=top_countries.index,
    y=top_countries.values
)

plt.title("Top 10 Countries by Hotel Bookings", fontsize=14)
plt.xlabel("Country")
plt.ylabel("Number of Bookings")
plt.xticks(rotation=45)

# Add value labels
for container in ax.containers:
    ax.bar_label(container)

plt.tight_layout()
plt.show()
```

### Observation

A small number of countries contribute a significant share of the total hotel bookings. This indicates that the hotel has strong demand from specific international markets.

### Business Insight

The hotel can prioritize marketing campaigns, promotional offers, and customer engagement strategies in countries with the highest booking volumes. At the same time, countries with lower booking volumes may represent opportunities for future market expansion.

### 10.3 Average Daily Rate (ADR) Analysis

Average Daily Rate (ADR) is one of the most important performance metrics in the hospitality industry. Comparing ADR across hotel types helps evaluate pricing strategies and understand revenue generation patterns.

**Business Question**

Which hotel type has the higher Average Daily Rate (ADR)?

```python
# Average Daily Rate by hotel type
adr_by_hotel = (
    df.groupby("hotel")["adr"]
      .mean()
      .sort_values(ascending=False)
)

adr_by_hotel
```

```python
plt.figure(figsize=(7,5))

ax = sns.barplot(
    x=adr_by_hotel.index,
    y=adr_by_hotel.values
)

plt.title("Average Daily Rate by Hotel Type", fontsize=14)
plt.xlabel("Hotel Type")
plt.ylabel("Average Daily Rate (ADR)")

# Add value labels
for container in ax.containers:
    ax.bar_label(container, fmt="%.2f")

plt.tight_layout()
plt.show()
```

### Observation

The comparison of Average Daily Rate (ADR) shows the difference in the average room price charged by each hotel type. This helps identify which hotel category follows a higher pricing strategy.

### Business Insight

Understanding ADR differences enables hotel management to evaluate pricing strategies and revenue performance across hotel types. Hotels with a higher ADR may generate greater revenue per booking, while lower-priced hotels may rely on higher booking volumes to maximize overall revenue.

### 10.4 Lead Time Analysis

Lead time represents the number of days between the booking date and the customer's arrival date. Analyzing lead time helps hotels understand booking behavior, forecast future demand, and improve pricing and operational planning.

**Business Question**

Which hotel type has the longer average lead time?

```python
# Average lead time by hotel type
lead_time_by_hotel = (
    df.groupby("hotel")["lead_time"]
      .mean()
      .sort_values(ascending=False)
)

lead_time_by_hotel
```

```python
plt.figure(figsize=(7,5))

ax = sns.barplot(
    x=lead_time_by_hotel.index,
    y=lead_time_by_hotel.values
)

plt.title("Average Lead Time by Hotel Type", fontsize=14)
plt.xlabel("Hotel Type")
plt.ylabel("Average Lead Time (Days)")

for container in ax.containers:
    ax.bar_label(container, fmt="%.1f")

plt.tight_layout()
plt.show()
```

### Observation

The analysis compares the average lead time across hotel types, highlighting differences in how far in advance customers typically make their reservations.

### Business Insight

Hotels with longer lead times can better forecast occupancy and optimize pricing strategies through early booking promotions. Understanding booking behavior also helps management improve inventory planning and operational efficiency.

### 10.10 Correlation Analysis

Correlation analysis measures the strength and direction of relationships between numerical features. Understanding these relationships helps identify patterns in customer behavior, pricing, booking characteristics, and operational metrics. It also supports feature selection and business decision-making.

**Business Question**

Which numerical features have the strongest relationships, and what do these relationships indicate about hotel booking behavior?

```python
# Select numerical features
numerical_df = df.select_dtypes(include=["int64", "float64"])

numerical_df.head()
```

```python
# Correlation matrix
correlation_matrix = numerical_df.corr()

correlation_matrix
```

```python
plt.figure(figsize=(16,12))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    linewidths=0.5
)

plt.title("Correlation Heatmap of Numerical Features", fontsize=16)

plt.tight_layout()
plt.show()
```

```python
# Top correlated feature pairs
corr_pairs = (
    correlation_matrix
    .unstack()
    .sort_values(ascending=False)
)

# Remove self-correlations
corr_pairs = corr_pairs[corr_pairs < 1]

corr_pairs.head(10)
```

### Business Insight

The correlation analysis indicates which operational and customer-related variables move together. Strong positive correlations can help hotel management understand related booking behaviors, while weak correlations suggest that multiple independent factors influence customer decisions. These insights are also valuable for feature selection in predictive modeling.
