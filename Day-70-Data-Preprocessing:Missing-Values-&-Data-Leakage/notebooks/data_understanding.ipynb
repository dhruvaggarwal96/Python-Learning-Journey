## Initial Business Hypotheses

1. Customers with month-to-month contracts may have a higher churn rate than customers with longer-term contracts.

2. Customers with shorter tenure may be more likely to churn than long-term customers.

3. Customers with higher monthly charges may have a higher probability of churn.

4. Customers without technical support may have a higher churn rate.

5. Customers using electronic check as their payment method may show different churn behavior compared with customers using other payment methods.

> These are initial hypotheses and will be tested using data during later stages of the project.

## Data Quality Findings

### 1. Customer ID

`customerID` contains a unique identifier for each customer. Since it does not represent meaningful customer behavior, it should not be used as a predictive feature.

### 2. Total Charges

`TotalCharges` represents the total amount charged to a customer but is currently stored as an object/string column.

Investigation found:

- 11 blank values are present.
- These values are blank strings rather than standard NaN values.
- All 11 customers with blank `TotalCharges` have `tenure = 0`.
- These records should be investigated before deciding how to handle the missing values.

### Initial Data Quality Considerations

- Convert `TotalCharges` to an appropriate numerical representation during preprocessing.
- Handle the blank `TotalCharges` values appropriately.
- Exclude `customerID` from predictive features.
- Avoid making preprocessing decisions using the test dataset.

## Potential Data Leakage Audit

Before model training, features should be checked for information that would not be available at the time of prediction.

### Target

`Churn` is the target variable and must not be included as an input feature.

### Customer ID

`customerID` will be excluded because it is a unique identifier and does not represent meaningful customer behavior.

### Potentially Safe Features

Features such as `Contract`, `PaymentMethod`, `tenure`, and `MonthlyCharges` represent customer information that would generally be available before a churn prediction is made.

### Total Charges

`TotalCharges` requires additional consideration because it represents accumulated customer charges. It should only be used if the value corresponds to information available at the prediction time.

### Leakage Prevention

Preprocessing operations such as imputation, scaling, encoding, and feature selection must be fitted using only the training data. The test set must remain unseen during preprocessing and model development.

> No model training is performed during the data-understanding stage.
