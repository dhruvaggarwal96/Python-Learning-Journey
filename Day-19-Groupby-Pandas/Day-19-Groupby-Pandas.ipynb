📅 Day 19 – GroupBy (Pandas)
groupby() is one of the most important functions in Pandas. It helps you analyze data category-wise.

📚 Learn
1️⃣ Total Sales by Category
df.groupby("Category")["Sales"].sum()
Example Output
Category
Furniture         50000
Office Supplies   70000
Technology        90000
2️⃣ Average Sales by Category
df.groupby("Category")["Sales"].mean()
Example Output
Category
Furniture         2500
Office Supplies   1800
Technology        3200
3️⃣ Count Records
df.groupby("Category").size()
Example Output
Category
Furniture         20
Office Supplies   40
Technology        28
4️⃣ Multiple Aggregations
df.groupby("Category")["Sales"].agg(["sum", "mean", "max", "min"])

Output:

                  sum   mean    max   min
Category
Furniture       50000  2500   9000   200
Technology      90000  3200  12000   500
💻 Practice Exercises

Use your existing datasets (sales_data.csv, customer_dataset.csv, amazon.csv) if applicable.

```python
import pandas as pd

df = pd.read_csv("sales_data.csv")

print(df.head())
```

```python
# Find the total sales amount for each product category.
df.groupby("Product_Category")["Sales_Amount"].sum()
```

```python
# Find the average sales amount for each product category.
df.groupby("Product_Category")["Sales_Amount"].mean()
```

```python
# Count the number of orders in each region.
df.groupby("Region").size()
```

```python
# Find the total quantity sold in each region.
df.groupby("Region")["Quantity_Sold"].sum()
```

```python
# Count the number of sales made by each sales representative.
df.groupby("Sales_Rep").size()
```

```python
# Find the average discount offered in each product category
df.groupby("Product_Category")["Discount"].mean()
```

```python
# Find the total sales amount for each sales channel.
df.groupby("Sales_Channel")["Sales_Amount"].sum()
```

```python
# Find the average quantity sold for each customer type.
df.groupby("Customer_Type")["Quantity_Sold"].mean()
```

```python
# Find the total sales amount by payment method.
df.groupby("Payment_Method")["Sales_Amount"].sum()
```

```python
# Find the average unit price for each product category.
df.groupby("Product_Category")["Unit_Price"].mean()
```

```python
# Find total sales amount and total quantity sold for each product category.
df.groupby("Product_Category")[["Sales_Amount", "Quantity_Sold"]].sum()
```

```python
# Find average sales amount and average discount by region.
df.groupby("Region")[["Sales_Amount", "Discount"]].mean()
```

```python
# Find the top 3 regions with the highest sales amount.
df.groupby("Region")["Sales_Amount"] \
  .sum() \
  .sort_values(ascending=False) \
  .head(3)
```

```python
# Find the sales representative with the highest total sales
df.groupby("Sales_Rep")["Sales_Amount"] \
  .sum() \
  .sort_values(ascending=False) \
  .head(1)
```

```python
# Find the payment method used the most.
df.groupby("Payment_Method") \
  .size() \
  .sort_values(ascending=False) \
  .head(1)
```

```python
# Find total sales amount for each combination of region and sales channel.
df.groupby(["Region", "Sales_Channel"])["Sales_Amount"].sum()
```

```python
# Find the average sales amount for each customer type in each region.
df.groupby(["Customer_Type", "Region"])["Sales_Amount"].mean()
```

```python
# Find the total sales amount generated by each sales representative in every region.
df.groupby(["Region", "Sales_Rep"])["Sales_Amount"].sum()
```

```python
# Find the product category with the highest sales amount.
df.groupby("Product_Category")["Sales_Amount"] \
  .sum() \
  .sort_values(ascending=False) \
  .head(1)
```

```python
# Find the region that sold the highest quantity of products.
df.groupby("Region")["Quantity_Sold"] \
  .sum() \
  .sort_values(ascending=False) \
  .head(1)
```
